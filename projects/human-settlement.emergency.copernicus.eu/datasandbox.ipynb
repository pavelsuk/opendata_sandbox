# Read data from geotiff

```python
import rasterio

fname = "data/GHS_POP_E2020_GLOBE_R2023A_4326_30ss_V1_0/GHS_POP_E2020_GLOBE_R2023A_4326_30ss_V1_0.tif"
```

## Read pixel values and metadata

```python

# Open the downloaded GeoTIFF file
with rasterio.open(fname) as src:
    print("Dimensions:", src.width, "x", src.height)
    print("Coordinate Reference System:", src.crs)
    print("Bounding Box:", src.bounds)

    # Read band 1 (population values) into a NumPy array
    pop_data = src.read(1)

    # Mask out NoData values (-200 or negative values)
    nodata = -200
    valid_pop = pop_data[pop_data != nodata]

    print("Total Global Population in file:", valid_pop.sum())
```

## Get population at a specific coordinate (Latitude / Longitude)

```python

# lat, lon = 50.0755, 14.4378  # Example: Prague, Czechia 7411.856648809795
lat, lon = 49.9633190171859, 13.929995317777179 # 1.45395284684168 Rozhledna Maminka

with rasterio.open(fname) as src:
    # Convert lat/lon (EPSG:4326) to pixel coordinates
    py, px = src.index(lon, lat)

    # Read value at pixel location
    population = src.read(1)[py, px]
    print(f"Population at ({lat}, {lon}):", population)
```
